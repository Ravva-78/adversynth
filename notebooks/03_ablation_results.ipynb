# Notebook 3 — Ablation Results
Compare classifier AUC-ROC across conditions: real-only vs real+synthetic.

```python
import matplotlib.pyplot as plt
import numpy as np

# Fill these in after running: python training/train_classifier.py --mode ablation
results = {
    'Real only':        {'Grade_0': 0.0, 'Grade_1': 0.0, 'Grade_2': 0.0, 'Grade_3': 0.0, 'Grade_4': 0.0},
    'Real + cGAN':      {'Grade_0': 0.0, 'Grade_1': 0.0, 'Grade_2': 0.0, 'Grade_3': 0.0, 'Grade_4': 0.0},
    'Real + CycleGAN':  {'Grade_0': 0.0, 'Grade_1': 0.0, 'Grade_2': 0.0, 'Grade_3': 0.0, 'Grade_4': 0.0},
}

grade_labels = ['Grade 0\nNo DR', 'Grade 1\nMild', 'Grade 2\nModerate', 'Grade 3\nSevere', 'Grade 4\nProlif.']
x = np.arange(5)
width = 0.25
colors = ['#378ADD', '#1D9E75', '#D85A30']

fig, ax = plt.subplots(figsize=(13, 6))
for i, (label, scores) in enumerate(results.items()):
    vals = [scores[f'Grade_{g}'] for g in range(5)]
    bars = ax.bar(x + i * width, vals, width, label=label, color=colors[i], alpha=0.85)

ax.set_xlabel('DR Grade', fontsize=11)
ax.set_ylabel('AUC-ROC', fontsize=11)
ax.set_title('Ablation: Classifier AUC-ROC by Grade\n(Higher is better — focus on Grade 3 & 4)', fontsize=12)
ax.set_xticks(x + width); ax.set_xticklabels(grade_labels)
ax.set_ylim([0, 1.05])
ax.axhline(0.8, color='red', linestyle='--', alpha=0.4, label='Target AUC=0.80')
ax.legend(fontsize=10)
ax.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.savefig('outputs/ablation_auc_bar.png', dpi=150)
plt.show()
```
