# Notebook 1 — Data Exploration
Explore APTOS 2019 class balance, image quality, and preprocessing results.

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
from pathlib import Path
from PIL import Image
import numpy as np

df = pd.read_csv('data/raw/train.csv')
print(f'Total samples: {len(df)}')
print(df['diagnosis'].value_counts().sort_index())
```

```python
# Class distribution bar chart
counts = df['diagnosis'].value_counts().sort_index()
labels = ['Grade 0\nNo DR', 'Grade 1\nMild', 'Grade 2\nModerate', 'Grade 3\nSevere', 'Grade 4\nProlif.']
colors = ['#378ADD', '#1D9E75', '#EF9F27', '#D85A30', '#A32D2D']

fig, ax = plt.subplots(figsize=(10, 5))
bars = ax.bar(labels, counts.values, color=colors, width=0.6)
for bar, count in zip(bars, counts.values):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 20,
            str(count), ha='center', va='bottom', fontsize=11)
ax.set_title('APTOS 2019 — Class Distribution', fontsize=13)
ax.set_ylabel('Number of images')
ax.axhline(len(df) * 0.05, color='red', linestyle='--', alpha=0.5, label='5% threshold')
ax.legend()
plt.tight_layout()
plt.savefig('outputs/class_distribution.png', dpi=150)
plt.show()
```

```python
# Sample grid — 3 images from each grade
fig, axes = plt.subplots(5, 3, figsize=(10, 16))
grade_names = ['No DR', 'Mild', 'Moderate', 'Severe', 'Proliferative']

for grade in range(5):
    grade_dir = Path(f'data/processed/train/{grade}')
    imgs = list(grade_dir.glob('*.png'))[:3]
    for col, img_path in enumerate(imgs):
        ax = axes[grade][col]
        ax.imshow(Image.open(img_path))
        ax.axis('off')
        if col == 0:
            ax.set_ylabel(f'Grade {grade}\n{grade_names[grade]}', fontsize=10)

fig.suptitle('Sample Preprocessed Images per Grade', fontsize=13)
plt.tight_layout()
plt.savefig('outputs/sample_grid.png', dpi=150)
plt.show()
```
